```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Dataset.csv')
print(f"Original dataset shape (Rows, Columns): {df.shape}\n")
```

```python
display(df.head())
```

```python
df.replace('Not Given', np.nan, inplace=True)
```

```python
duplicate_count = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_count}")
```

```python
missing_values = df.isnull().sum()
print("Number of missing values per column:\n", missing_values[missing_values > 0], "\n")
```

```python
df.drop_duplicates(inplace=True)
```

```python
df['director'] = df['director'].fillna('Unknown')
```

```python
if not df['country'].mode().empty:
    most_common_country = df['country'].mode()[0]
    df['country'] = df['country'].fillna(most_common_country)
```

```python
df.dropna(subset=['date_added', 'rating', 'duration'], inplace=True)
print(f"Dataset shape after handling null values: {df.shape}\n")
```

```python
df['date_added'] = df['date_added'].str.strip()
df['date_added'] = pd.to_datetime(df['date_added'], format='mixed')
```

```python
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month
```

```python
df['type'] = df['type'].astype('category')
```

```python
df['duration_num'] = df['duration'].str.extract(r'(\d+)').astype(float)
```

```python
print("Data transformation successful. New columns have been added.\n")
display(df[['date_added', 'year_added', 'month_added', 'duration', 'duration_num']].head())
```

```python
clean_filename = 'cleaned_Dataset.csv'
df.to_csv(clean_filename, index=False)
print(f"Cleaned dataset successfully saved as '{clean_filename}'.")
```
