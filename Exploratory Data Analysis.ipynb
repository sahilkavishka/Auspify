```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv('cleaned_Dataset.csv')
```

```python

print(df.info())
print("\n Summary Statistics:")
display(df.describe(include='all'))
```

```python
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x='type', palette='Set2')
plt.title('Distribution of Content Types (Movies vs TV Shows)', fontsize=14)
plt.xlabel('Content Type')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
# Using value_counts to get the top 10 countries
top_countries = df['country'].value_counts().head(10)
sns.barplot(y=top_countries.index, x=top_countries.values, palette='viridis')
plt.title('Top 10 Countries with Most Netflix Content', fontsize=14)
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
top_genres = df['listed_in'].value_counts().head(10)
sns.barplot(y=top_genres.index, x=top_genres.values, palette='magma')
plt.title('Top 10 Content Categories (Genres)', fontsize=14)
plt.xlabel('Number of Titles')
plt.ylabel('Category')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
# Filter data to exclude outliers/errors in years if any
trend_data = df[df['year_added'] >= 2010]
sns.countplot(data=trend_data, x='year_added', hue='type', palette='muted')
plt.title('Content Added to Netflix Over the Years (2010 onwards)', fontsize=14)
plt.xlabel('Year Added')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.legend(title='Type')
plt.show()
```
